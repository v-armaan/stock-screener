```python
import yfinance as yf
import pandas as pd
import numpy as np
from yahooquery import search
```

```python
a="S&P 500"
```

```python
search_results=search(a)
search_results
```

```python
op_search_results=[]
count=0
for i in search_results["quotes"]:
    if i["quoteType"]=="EQUITY" and count<5:
        op_search_results.append(i)
        count+=1
           
    
```

```python
op_search_results[0]
```

CODE FOR PRINTING LIST OF TOP 5 COMPANIES AFTER SEARCH AND TAKING INPUT

```python
serial=0
for i in op_search_results:
    serial=serial+1
    print(serial,". ",i["shortname"],"(",i["symbol"],")")
choice=1
print("chosen company=",op_search_results[choice-1]["shortname"],"(",op_search_results[choice-1]["symbol"],")")
```
