```python
fig=make_subplots(rows=2,cols=1,shared_xaxes=True,vertical_spacing=0.05,row_heights=[0.75,0.25])
#candlestick
fig.add_trace(go.Candlestick(x=df.index,open=df["Open"],high=df["High"],low=df["Low"],close=df["Close"],name="Price"),row=1,col=1)
#sma
ma=calculations.calculate_moving_averages(df)
fig.add_trace(go.Scatter(x=df.index,y=ma["SMA_20"],name="SMA 20",mode="lines"),row=1,col=1)
#ema
fig.add_trace(go.Scatter(x=df.index,y=ma["EMA_20"],name="EMA 20",mode="lines"),row=1,col=1)
#volume
volume_colors=[]
for i in range(len(df)):
    if df["Close"].iloc[i]>df["Open"].iloc[i]:
        volume_colors.append("green")
    else:
        volume_colors.append("red")
fig.add_trace(go.Bar(x=df.index,y=df["Volume"],name="Volume",marker_color=volume_colors),row=2,col=1)
fig.update_layout(
    height=800,
    xaxis_rangeslider_visible=False,
    plot_bgcolor="black"
)
fig.update_xaxes(
    showline=False,
    mirror=False
)

fig.show()

```
